```python
#with function
import statistics
data=[1,1,1,2,2,3,3,4]

def cal_mean(data):
  return statistics.mean(data)
def cal_median(data):
  return statistics.median(data)
def cal_mode(data):
  return statistics.mode(data)
def cal_stdev(data):
  return statistics.stdev(data)
def cal_variance(data):
  return statistics.variance(data)

print(f"Mean:{cal_mean(data)}")
print(f"Median:{cal_median(data)}")
print(f"Mode:{cal_mode(data)}")
print(f"Standard Deviation:{cal_stdev(data)}")
print(f"Variance:{cal_variance(data)}")


```
